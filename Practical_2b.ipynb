## 1) What is a traceback? <a name="Part1"></a>

When Python cannot run your code it stops and prints an error message called a **traceback**. A traceback can look intimidating, but it always contains the same three pieces of information:

 1. **Where** the problem happened: the line number, marked with an arrow <code>----></code> or with <code>^^^^</code> underneath the offending code.
 2. **What kind** of problem it is: the *error type*, such as <code>NameError</code> or <code>TypeError</code>.
 3. **Why**: a short message after the error type, such as <code>name 'rainfal' is not defined</code>.

<b>The golden rule: read a traceback from the bottom up.</b> The last line tells you what went wrong. Then look above it to find the line that caused it.

Let's look at an example. The code below converts a temperature from Kelvin to degrees Celsius. Run the code box and look at the traceback.

```python
# Imagine we have read this value from a text file of weather station data.
# Values read from text files often arrive as strings, not numbers!
station_temperature = "288.4"

temperature_C = station_temperature - 273.15
print("The temperature is", temperature_C, "degrees C")
```

Reading this from the bottom up:

 - The last line says <code>TypeError: unsupported operand type(s) for -: 'str' and 'float'</code>. So we tried to subtract (<code>-</code>) a <code>float</code> from a <code>str</code> (a string of text). As we saw in Practical 1, Python treats text and numbers as different data types, and it does not know how to subtract a number from a word.
 - Just above, the arrow points at the line <code>temperature_C = station_temperature - 273.15</code>. That is where Python gave up.

Notice that the line where the error *appears* is not always where the *cause* is. The real problem is two lines higher up: <code>station_temperature</code> is the string <code>"288.4"</code> rather than the number <code>288.4</code>. We can fix it by converting it with <code>float()</code>, which you met in Practical 1:

```python
station_temperature = "288.4"
station_temperature = float(station_temperature) # convert the string to a floating point number

temperature_C = station_temperature - 273.15
print("The temperature is", temperature_C, "degrees C")
```

<div class="alert alert-block alert-info">
<b>Tip: when the traceback is long.</b> Later in the course, when you write your own functions (Practical 3) and use bigger packages such as Pandas (Practical 6), tracebacks will get longer. They will show a chain of lines, one for each function that was running when the error happened. The golden rule still applies: start at the bottom with the error type and message, then work upwards until you reach a line of <i>your own</i> code. That is usually the line you need to fix.

## 2) A field guide to common errors <a name="Part2"></a>

You will meet the following error types again and again. It is worth getting to know them:

| Error type | What it usually means | Earth-science flavoured example |
|---|---|---|
| <code>SyntaxError</code> | Python cannot even read your code: a missing colon, bracket or quote | <code>for year in years</code> (no <code>:</code> at the end) |
| <code>IndentationError</code> | The spaces at the start of a line are wrong | The line under an <code>if</code> statement is not indented |
| <code>NameError</code> | You used a variable name that does not exist (often a typo, or a cell you have not run yet) | <code>print(rainfal)</code> instead of <code>print(rainfall)</code> |
| <code>TypeError</code> | You used a value of the wrong *type*, e.g. text where a number is needed | <code>"1085" * 0.0065</code> |
| <code>IndexError</code> | You asked for a position in a list that does not exist | Asking for <code>magnitudes[5]</code> in a list of 5 earthquakes |
| <code>KeyError</code> | You asked a dictionary for a key it does not have (keys are case sensitive!) | <code>vei['etna']</code> when the key is <code>'Etna'</code> |
| <code>ValueError</code> | The type is right but the value is not allowed | <code>float("12.3 mm")</code> |
| <code>ZeroDivisionError</code> | You divided by zero | Mean rainfall over a period with 0 days |

A <code>SyntaxError</code> is a little different from the others: Python finds it *before* running anything, so none of the code in that box will run, not even the lines above the mistake. Try it below. Can you spot the problem from the <code>^</code> marker?

```python
eruption_years = [1669, 1928, 1981, 2002]
print("Checking the list of Etna eruptions...") # This line will NOT run, because of the SyntaxError below

for year in eruption_years
    print("Etna erupted in", year)
```

Python points its <code>^</code> marker at the end of the <code>for</code> line and tells us it <code>expected ':'</code>. Every <code>for</code>, <code>if</code>, <code>elif</code> and <code>else</code> line must end with a colon.

```python
# (a) Mean monthly rainfall in Manchester [mm] (approximate long-term values)
#------'INSERT CODE HERE'------
# Error type:
# Line:
# Why:
#------------------------------
rainfall = [79, 59, 63, 52, 58, 69, 66, 79, 75, 97, 92, 95]

total = 0
for month in rainfall:
    total = total + month

mean_rainfall = total / len(rainfal)
print("Mean monthly rainfall =", round(mean_rainfall, 1), "mm") # Expected answer: 73.7 mm
```

```python
# (b) Air pressure at the summit of Snowdon using the International Standard Atmosphere
# p = 101325 * (1 - 2.25577e-5 * h)**5.25588, with h in metres and p in Pascals
#------'INSERT CODE HERE'------
# Error type:
# Line:
# Why:
#------------------------------
summit_height = "1085" # metres, as read from a text file

pressure = 101325 * (1 - 2.25577e-5 * summit_height)**5.25588
print("Pressure at the summit =", round(pressure / 100, 1), "hPa") # Expected answer: 889.5 hPa
```

```python
# (c) Print the magnitude of each earthquake in a list, one per line
#------'INSERT CODE HERE'------
# Error type:
# Line:
# Why:
#------------------------------
magnitudes = [4.2, 5.1, 3.8, 6.4, 4.9]

for i in range(1, len(magnitudes) + 1):
    print("Earthquake", i, "had magnitude", magnitudes[i])
# Expected answer: five lines, starting with 'Earthquake 1 had magnitude 4.2'
```

```python
# (d) Look up the Volcanic Explosivity Index (VEI) of a famous eruption
#------'INSERT CODE HERE'------
# Error type:
# Line:
# Why:
#------------------------------
vei = {'Vesuvius 79 AD': 5, 'Krakatau 1883': 6, 'Pinatubo 1991': 6, 'Etna 2002': 2}

eruption = 'pinatubo 1991'
print("The VEI of", eruption, "was", vei[eruption]) # Expected answer: 6
```

```python
# (e) Classify eruptions as 'small' (VEI below 4) or 'large' (VEI of 4 or more)
#------'INSERT CODE HERE'------
# Error type:
# Line:
# Why:
#------------------------------
vei_values = [2, 6, 3, 5, 1]

for value in vei_values:
    if value < 4:
    print("VEI", value, "is a small eruption")
    else:
        print("VEI", value, "is a large eruption")
# Expected answer: five lines, starting with 'VEI 2 is a small eruption'
```
